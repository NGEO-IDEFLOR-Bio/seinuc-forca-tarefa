```python
import os

def listar_arquivos_projeto(caminho_projeto):
    print(f"Mapeando diretório: {caminho_projeto}\n")
    print(f"{'NOME DO ARQUIVO':<50} | {'TAMANHO (KB)':<12}")
    print("-" * 65)

    if not os.path.exists(caminho_projeto):
        print("Erro: O caminho especificado não existe.")
        return

    for raiz, pastas, arquivos in os.walk(caminho_projeto):
        # Calcula o nível de indentação para pastas
        nivel = raiz.replace(caminho_projeto, '').count(os.sep)
        indentacao = '  ' * nivel
        
        # Mostra a pasta atual
        if raiz != caminho_projeto:
            print(f"{indentacao}[Pasta] {os.path.basename(raiz)}")

        # Lista os arquivos
        for nome in arquivos:
            caminho_completo = os.path.join(raiz, nome)
            tamanho = os.path.getsize(caminho_completo) / 1024  # Converte para KB
            
            # Formata a exibição
            prefixo = '  ' * (nivel + 1)
            print(f"{prefixo}{nome:<{50 - len(prefixo)}} | {tamanho:>10.2f} KB")

# Ajuste o caminho abaixo se necessário
caminho = r"H:\Meu Drive\IDEFLOR\SEINUC\producao\abnt"
listar_arquivos_projeto(caminho)
```
